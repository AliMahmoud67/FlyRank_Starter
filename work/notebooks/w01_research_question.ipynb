```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print(f"Starter dataset: {len(df):,} content items × {len(df.columns)} columns")

# Pages with meaningful visibility but low CTR.
low_ctr_visible = (
    (df["impressions_90d"] >= 500)
    & (df["avg_position"] > 0)
    & (df["avg_position"] <= 20)
    & (df["ctr"] < 0.5)
)

# Older pages ranking on page one.
older_page_one = (
    (df["avg_position"] > 0)
    & (df["avg_position"] <= 10)
    & (df["content_age_days"] >= 180)
)

# Pages satisfying both conditions.
review_candidate = low_ctr_visible & older_page_one

print(f"Low-CTR visible pages: {low_ctr_visible.sum():,} ({low_ctr_visible.mean():.1%})")
print(f"Older page-one pages: {older_page_one.sum():,} ({older_page_one.mean():.1%})")
print(f"Overlap of both conditions: {review_candidate.sum():,} ({review_candidate.mean():.1%})")
```
