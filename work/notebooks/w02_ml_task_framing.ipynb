## 3. Success metric

*One metric you can defend. What number means 'good'?*

**Precision@K** measures the proportion of the top K ranked pages that are actually worthwhile review candidates.

For example, if the model ranks 50 pages at the top and 30 of those pages meet the validated definition of a worthwhile review candidate, then Precision@50 is 30/50 = 0.60.

A higher Precision@K means that a larger proportion of the pages placed at the top of the review queue are relevant candidates
